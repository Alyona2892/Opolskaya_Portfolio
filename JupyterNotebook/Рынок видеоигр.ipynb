```python
# Комментарий ревьюера
def show_missing_stats(tmp0):
    """
    Функция для отображения статистики пропущенных значений в DataFrame.
    """
    missing_stats = pd.DataFrame({
        'Кол-во пропусков': tmp0.isnull().sum(),
        'Доля пропусков': tmp0.isnull().mean()
    })
    missing_stats = missing_stats[missing_stats['Кол-во пропусков'] > 0]
    
    if missing_stats.empty:
        return "Пропусков в данных нет"
    
    # Форматируем при выводе через Styler
    return (missing_stats.style.format({'Доля пропусков': '{:.4f}'}).background_gradient(cmap='coolwarm'))
show_missing_stats(games)
```

А вот с пропусками в оценках пользователей и критиков сложнее - именно по ним будет проводиться категоризация. Для принятия решения о том, что делать с пропусками, рассмотрим как пропуски распределяются по годам.

```python
# Группируем пропуски critic_score по годам и считаем количество 
no_critic_grouped_year = (
    games[games['critic_score'].isna()]
    .groupby('year_of_release')['name']
    .count()
)
print(f'Количество пропусков в столбце critic_score:')
print(no_critic_grouped_year)
# Группируем пропуски user_score по годам и считаем количество 
no_user_grouped_year = (
    games[games['user_score'].isna()]
    .groupby('year_of_release')['name']
    .count()
)
print(f'Количество пропусков в столбце user_score:')
print(no_user_grouped_year)
```

В резуьтате мы видим, что на интересующий нас период (2000-2013 гг) приходится большое количество пропусков, поэтому заменять пропуски средним значением не корректно в виду необходимости дальнейшей категоризации игр по оценкам. Это сильно повлияет на резуьтаты категоризации и анализа, в связи с чем решаем оставить данные как есть и в дальнейшем при категоризации выделять отдельную группу, где отсутствуют нужные данные.
