У нас есть еще 2 столбца типо object: founded_month и founded_quarter. При этом столбец founded_at содержит полную дату и скорее всего founded_month и founded_quarter - производные от него. Проверим, чтобы понять есть ли у нас строки, где известно founded_at, но отсутствуют month и quarter. Если да, мы сможем их заполнить. 

```python
# Выделяем, где есть founded_at, нет founded_month и нет founded_quarter
mask_founded_at = invest_info['founded_at'].notna()
mask_no_month = invest_info['founded_month'].isna()
mask_no_quarter = invest_info['founded_quarter'].isna()
# Считаем пропуски founded_month и founded_quarter
no_month = mask_no_month.sum()
no_quarter = mask_no_quarter.sum()
# Считаем где есть founded_at и нет founded_month/founded_quarter
no_month_with_date = (mask_founded_at & mask_no_month).sum()
no_quarter_with_date = (mask_founded_at & mask_no_quarter).sum()
# То же самое,но в доле (%)
share_no_month = no_month_with_date / no_month
share_no_quarter = no_quarter_with_date / no_quarter

print(f'Пропусков в founded_month: {no_month}')
print(f'Из них есть полная дата: {no_month_with_date}, {share_no_month:.2%}')
print(f'Пропусков в founded_quarter: {no_quarter}')
print(f'Из них есть полная дата: {no_quarter_with_date}, {share_no_quarter:.2%}')
```

Очень малое количество данных содержит полную дату, на основе которой можно заполнить пропуски в month и quarter. Заполнение не обязательно, т к не окажет сильного влияние на дальнейшее исследование, но мы решаем все же сделать заполнение.  

```python
# Заполняем месяц и квартал на основе полной даты
mask_fill_month = (invest_info['founded_at'].notna() & invest_info['founded_month'].isna())
mask_fill_quarter = (invest_info['founded_at'].notna() & invest_info['founded_quarter'].isna())
invest_info.loc[mask_fill_month, 'founded_month'] = (invest_info.loc[mask_fill_month, 'founded_at'].dt.to_period('M').astype(str))
# Для quarter учитываем формат, в котором представлены денные (YYYY-QN)
invest_info.loc[mask_fill_quarter, 'founded_quarter'] = (invest_info.loc[mask_fill_quarter, 'founded_at'].dt.year.astype(str)
    + '-Q' + invest_info.loc[mask_fill_quarter, 'founded_at'].dt.quarter.astype(str))
# Проверим, что все заполнилось и ничего не потерялось
month_quarter_non_null = pd.DataFrame({
    'non_null_old': invest_info_old[['founded_month', 'founded_quarter']].count(),
    'non_null_new': invest_info[['founded_month', 'founded_quarter']].count()
})
month_quarter_non_null['added'] = (month_quarter_non_null['non_null_new'] - month_quarter_non_null['non_null_old'])
month_quarter_non_null
```

Определите компании с аномальным объёмом общего финансирования — используйте метод IQR отдельно по каждому сегменту. Напомним, что все нишевые сегменты должны быть объединены в одну группу, а средние — в другую.

Определите сегменты рынка с наибольшей долей компаний, получивших аномальное финансирование, и выведите топ таких сегментов.

```python
# Считаем границы и отмечаем True компании с аномальным финансированием
all_markets = invest_info['market'].unique()
invest_info['out_of_bound'] = False
for segment in all_markets:
    market_data = invest_info[invest_info['market'] == segment]
    q1 = market_data['funding_total_usd_mln'].quantile(0.25)
    q3 = market_data['funding_total_usd_mln'].quantile(0.75)
    iqr = q3-q1
    lower_bound = q1 - 1.5*iqr
    upper_bound = q3 + 1.5*iqr
    invest_info.loc[(invest_info['market'] == segment) & ((invest_info['funding_total_usd_mln']>upper_bound) | (invest_info['funding_total_usd_mln']<lower_bound)), 'out_of_bound'] = True
# Смотрим сколько компаний получилось    
print(invest_info['out_of_bound'].value_counts())
print(f"Доля компаний с аномальными инвестициями: {invest_info['out_of_bound'].mean():.2%}")
```

В датасете есть 5203 компаний, объем инвестиций которых является аномальным в своем сегменте. Это составляет 12,7% от общего числа компаний. 

```python
# Считаем долю компаний с аномальным финансированием по сегментам (%)
out_of_bound_share = invest_info.groupby('market')['out_of_bound'].mean()
out_of_bound_share = round((out_of_bound_share*100),2)
# Найдем 75-й перцентиль для долей
p75_out_share = out_of_bound_share.quantile(0.75)
# За сегменты с наибольшей долей примем все сегменты, где доля выше 75-го перцентиля
high_out_share = out_of_bound_share[out_of_bound_share > p75_out_share]
# Отсортируем по убыванию доли
high_out_share.sort_values(ascending=False)
```

Мы отобрали сегменты, в которых доля компаний, получивших аномальное финансирование, наиболее высока (выше 75-го персентиля). ### 4.3 Годовая динамика доли возвращённых средств по типам финансирования

Заказчик хочет знать, какая часть вложенных или выданных денег со временем возвращается обратно инвесторам или финансистам. Ваша цель — для каждого года и каждого вида финансирования рассчитать нормированные значения возврата средств: то есть какую долю возвращённые средства составляют от предоставленных. При этом слишком большие аномальные значения, то есть неадекватные выбросы, нужно заменить на пропуски.

Совет: когда будете делить сумму возвращённых средств на суммарный объём привлечённого финансирования по конкретному году, добавьте к знаменателю небольшое число, например `1e-60`. Это поможет избежать деления на ноль.

```python
# Собираем сводную таблицу для invest_info (в млн USD)
funding_type_year = (invest_info_actual.pivot_table(
    index='mid_funding_year',
    values=finance_types,
    aggfunc='sum')/1000000) 
funding_type_year = funding_type_year.round(2)
funding_type_year
```

```python
# Собираем сводную таблицу для returns_info
returns_type_year = (returns_info.pivot_table(
    index='year',
    values=finance_types,
    aggfunc='sum')) 

returns_type_year = returns_type_year.round(2)
returns_type_year
```

Доля может быть выше 100% - это говорит о прибыли инвестора в этому году (вернул больше, чем вложил), поэтому аномальными признаем значения, когда изначально вложенных денег 0, а возврат > 0.

```python
# Считаем долю возвращенных средств от предоставленных (в %)
returns_share = round(returns_type_year/ (funding_type_year + 1e-60) *100, 2)
# Меняем аномальные значения на пропуски 
normal_returns_share = returns_share.where(((funding_type_year==0) & (returns_type_year>0))==False)
normal_returns_share
```

Постройте график, на котором отобразите нормированные значения возврата средств для типов финансирования `venture`, `debt_financing`, `private_equity`, `seed` и `angel`.

Сделайте вывод о том, в каких типах финансирования наблюдается наиболее устойчивый рост показателя.

```python
# Отбираем нужные для графика типы финансирования
types_for_plot = ['venture', 'debt_financing', 'private_equity', 'seed', 'angel']
share_for_plot = normal_returns_share[types_for_plot]
# Строим линейный график
share_for_plot.plot(figsize=(14,9), colormap='tab20', marker='o')

plt.grid(axis='y')
plt.legend(title='Тип финансирования', loc='upper right')
plt.ylabel('Доля возвращенных средств (%)')
plt.xlabel('')
plt.title('Доли возврата средств по годам и типам финансирования')
plt.show()
```

Наиболее стабильна доля возвращенных средств у debt_financing - чаще всего доля возврата находится в диапозоне 40-100%. Скорее всего это связано с самим типом финансирования - такие долговые обязательства, как, например, заем в банке, более предсказуемы для компании с точки зрения распределния финансов, а также предполагают четкий график погашения долга. Если рассмотреть private_equity, то тоже можно сказать, что с 2005 года колебания не очень резкие - диапозон доли примерно от 50 до 100% (не считая 2014 года, где у всех типов финансирования заметен резкий скачок). Венчурные инвестиции не так предсказуемы в вопросе возврата средств - резкие скачки в начале периода, затем упадок доли возврата и стагнация с 2006 по 2013 год. 

Вообще, возврат инвестиций ведь может быть отложенным и на мой взгляд здесь важно посмотреть и кумулятивные суммы, чтобы понять, что к такому-то году было возвращено столько-то вложенных (совокупно к этому году) инвестиций.

```python
# Считаем кумулятивные суммы
funding_type_year_cum = funding_type_year.cumsum()
returns_type_year_cum = returns_type_year.cumsum()
# Считаем долю возвращенных средств от предоставленных (в %)
returns_share_cum= round(returns_type_year_cum/ (funding_type_year_cum + 1e-60) *100, 2)
returns_share_cum
```

```python
# Отбираем нужные для графика типы финансирования
types_for_plot_cum = ['venture', 'debt_financing', 'private_equity', 'seed', 'angel']
share_for_plot_cum = returns_share_cum[types_for_plot_cum]
# Строим линейный график
share_for_plot_cum.plot(figsize=(14,9), colormap='tab20', marker='o')

plt.grid(axis='y')
plt.legend(title='Тип финансирования', loc='upper right')
plt.ylabel('Кумулятивная доля возвращенных средств (%)')
plt.xlabel('')
plt.title('Кумулятивная доля возврата средств по годам и типам финансирования')
plt.show()
```

С точки зрения кумулятивной доли возврата, private_equity можно назвать стабильным типом финансирования, начиная с 2004 года колебания не высокие, и к 2014 году было возвращено уже около 80% от вложенных инвестиций. Angel стабилен, в начале периода хороший рост, но начиная с 2008 года такой динамики роста уже не наблюдается. Однако около 60% финансирование и angel, и debt_financing уже были возвращены к концу периода. и  А вот venture отличается спадом в 2005-2006 годах и далее есть длительная стагнация в доле возврата средств, что говорит о том, что разрыв между вложениями и возвратом стал выше, чем был в начале периода. У типа финансирования seed динамика отрицательная - с самого начала и до конца периода больше заметна тенденция спада, кумулятивная доля возврата не особо растет и не отличается стабильностью.
